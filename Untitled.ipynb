# Part 2: Deep-Dive Analysis

## Customer & Sales Segmentation Analysis

### Objective

The objective of this analysis is to segment customers based on their total sales contribution and identify high-value, medium-value, and low-value customer segments.

# Customer-level sales calculate:-

```python
customer_sales = df.groupby(
    ["Customer_ID", "Customer_Name"]
)["Total_Sales"].sum().reset_index()

customer_sales.head()
```

# Customer sales check karo:-

```python
customer_sales.shape
```

```python
customer_sales["Total_Sales"].describe()
```

# Customer segments create :-

```python
# HIGH VALUE
# MEDIUM VALUE
# LOW VALUE
```

```python
q1=customer_sales["Total_Sales"].quantile(0.33)
q2=customer_sales["Total_Sales"].quantile(0.66)

def assign_segments(sales):
    if sales >=q2:
        return "High Value"
    elif sales >=q1:
        return "Medium Value"
    else:
        return "Low Value"
customer_sales["Customer_segment"]=customer_sales["Total_Sales"].apply(assign_segments)

customer_sales.head(5)
        
        
```

```python
customer_sales["Customer_segment"].value_counts()
```

# segement-wise revenue:-

```python
segment_rev=customer_sales.groupby("Customer_segment")["Total_Sales"].agg(["count","sum","mean"]).reset_index()
segment_rev
```

```python
# sort from of segment revenue:-
```

```python
segment_revenue=segment_rev.sort_values("sum",ascending=False)
segment_revenue
```

# Segment visualization:-

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.bar(segment_rev["Customer_segment"],segment_rev["sum"])

plt.xlabel("customer segment")
plt.ylabel("total revenue")
plt.title("revenue by customers segment")
plt.show()
```

# Top 10 Customers:-

```python
top_customers=customer_sales.sort_values("Total_Sales", ascending=False).head(10)
top_customers
```

```python
plt.figure(figsize=(8,5))
plt.barh(top_customers["Customer_Name"],top_customers["Total_Sales"])

plt.xlabel("total sales")
plt.ylabel("customers")
plt.title(" top 10 customers by revenue")
plt.gca().invert_yaxis()
plt.show()
```

# City-level deep dive:-

# Top 10 cities:-

# char of top 10 cities:-

```python
plt.figure(figsize=(8,5))
plt.bar(top_cities["City"],top_cities["Total_Revenue"])
plt.xlabel("total revenue")
plt.ylabel("city")
plt.title("top 10 total revenue by city")
plt.show()
```

# Product-level deep dive:-

```python
product_analysis=df.groupby("Product").agg(Total_Revenue=("Total_Sales","sum"),Total_Quantity=("Quantity","sum"),Average_Unit_Price=("Unit_Price","mean")).reset_index()
product_analysis=product_analysis.sort_values("Total_Revenue",ascending=False)
product_analysis
```

# Top 10 products chart:-

```python
top_product=product_analysis.head(10)
plt.figure(figsize=(8,5))
plt.bar(top_product["Product"],top_product["Total_Revenue"])
plt.xlabel("Total Revenue")
plt.ylabel("product")
plt.title("Top 10 product by Total Revenue")
plt.show()
```

# Gender-wise sales analysis:-

```python
gender_analysis=df.groupby("Gender").agg(Total_Revenue=("Total_Sales","sum"),Total_Orders=("Order_ID","nunique"),Total_Quantity=("Quantity","sum")).reset_index()
gender_analysis
```

# Age group analysis:-

```python
df["Age_Group"]=pd.cut(df["Age"],bins=[17, 25, 35, 45, 55, 65],labels=["18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65"
    ]
)
age_analysis=df.groupby("Age_Group",observed=False)["Total_Sales"].sum().reset_index()
age_analysis
```

```python
plt.figure(figsize=(8,5))
plt.bar(age_analysis["Age_Group"].astype(str),age_analysis["Total_Sales"])
plt.xlabel("Age Group")
plt.ylabel("Total sales")
plt.title("Revenue by age group")
plt.show()
        
```
